```python
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.cluster import AgglomerativeClustering

import numpy as np

# Načtení souboru
file_path = "/Users/zphilipp/git/suggest/data/query_by_division.txt.1"
df = pd.read_csv(file_path, sep="\t", header=None)
df.columns = ['division', 'query', 'lema', 'value']
df['value'] = pd.to_numeric(df['value'], errors='coerce')
df = df.dropna(subset=['value'])

# Seskupení dle query
grouped = df.groupby(['lema'])['value'].sum().reset_index()

# Inicializace embedding modelu
model = SentenceTransformer('all-MiniLM-L6-v2')

# Výpočet embeddingu pro každý query
grouped['embedding'] = grouped['lema'].apply(lambda x: model.encode(x))

```

```python
# Výpis embeddingu pro první 3 query (pro ukázku)
for _, row in grouped.head(3).iterrows():
    print(f"Query: {row['lema']}")
    print(f"Embedding: {np.array2string(row['embedding'], precision=3)}\n")
```

```python
# Parametry
similarity_threshold = 0.9
distance_threshold = 1 - similarity_threshold

# Převod embeddingů na matici
embeddings = np.vstack(grouped['embedding'].values)

# Shlukování pomocí kosinové vzdálenosti
clustering_model = AgglomerativeClustering(
    n_clusters=None,
    metric='cosine',
    linkage='average',
    distance_threshold=distance_threshold
)
labels = clustering_model.fit_predict(embeddings)

# Výsledek
grouped['cluster'] = labels
```

```python
# Výpis: všechny clustery s víc než jedním dotazem včetně hodnoty
for cluster_id in sorted(grouped['cluster'].unique()):
    cluster_df = grouped[grouped['cluster'] == cluster_id]
    if len(cluster_df) > 0:
        print(f"\nCluster {cluster_id} ({len(cluster_df)} queries):")
        for _, row in cluster_df.iterrows():
            print(f" - {row['lema']}  (value = {row['value']:.0f})")
```
