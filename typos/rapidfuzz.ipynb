```python
from rapidfuzz import process, fuzz
from collections import defaultdict

phrases = {}
with open("data/phrases.txt", "r", encoding="utf-8") as f:
    for line in f:
        *phrase_parts, count = line.strip().split()
        phrase = " ".join(phrase_parts)
        try:
            phrases[phrase] = int(count)
        except ValueError:
            continue

phrases_by_wordcount = defaultdict(list)
for phrase in phrases.keys():
    count = len(phrase.split())
    phrases_by_wordcount[count].append(phrase)

def correct_fuzzy_by_wordcount(input_query: str, min_score: float = 87.0) -> str:
    input_query = input_query.lower().strip()
    word_count = len(input_query.split())

    candidates = phrases_by_wordcount.get(word_count, list(phrases.keys()))

    match = process.extractOne(
        input_query,
        candidates,
        scorer=fuzz.token_sort_ratio
    )

    if match and match[1] >= min_score:
        return match[0]
    else:
        return input_query
        
# Test
test_queries = [
    "massge",
    "grate wolf lodge",
    "costoco",
    "pedicuer",
    "phpto shoot",
    "massage hwights couples",
    "sam’s club",
    "laser hair remuval",
    "laser hair remo",
    "pizzza",
    "piza restaurant",
    "oil ch",
    "couples ma",
    "cost",
    "coast",
    "escape rom",
    "laser heir",
    "padicure",
    "gol",
    "hed spa",
    "zom",
    "zoom",
    "foot",
    "mail",
    "mad",
    "amc"
]

print("Test:\n")
for query in test_queries:
    corrected = correct_fuzzy_by_wordcount(query)
    print(f"'{query}' → '{corrected}'")
```
